# Python Statements And Comments 
<div style="font-size:20px">
    
**Statements:** A statement is an instruction that a Python interpreter can execute.

So, in simple words, we can say anything written in Python is a statement.

**Note:** Your lines of code must be syntactically correct, otherwise they are not valid statements. 

**Comments:** Comments in Python are the lines in the code that are ignored by the interpreter 
during the execution of the program.

## Single-line Comment
<div style="font-size:20px">
In Python, single-line comments are indicated by a hash sign(#). 
The interpreter ignores anything written after the # sign, and it is effective till the end of the line.
</div>

## Multi-Line Comments
<div style="font-size:20px">
In Python, there is no separate way to write a multi-line comment. 
Instead, we need to use a hash sign at the beginning of each comment line to make it a multi-line comment

</div>

Conditional statements in Python are used to execute certain blocks of code based on specific conditions.

## Nesting if Statements: Layered Decisions
<div style="font-size:20px">
Concept: Place one conditional inside another to handle complex scenarios.

```python
# Example
num = -3
if num < 0:
    if num % 2 == 0:
        print("Negative even number")
    else:
        print("Negative odd number")
else:
    print("Non-negative number")
```

```python
n = int(input("Enter a number: "))

if n >= 0:                            # outer check: sign
    if n % 2 == 0:                    # inner check: parity
        print("Positive even")
    else:
        print("Positive odd")
else:
    if n % 2 == 0:
        print("Negative even")
    else:
        print("Negative odd")
```

```python
# Login with Role-Based Access 
username = input("Username: ")
password = input("Password: ")

if username == "tejal":
    if password == "tejal123":
        print("Access granted: Admin dashboard")
    else:
        print("Wrong password for admin")
else:
    if username == "guest":
        if password == "guest123":
            print("Access granted: Guest mode")
        else:
            print("Wrong password for guest")
    else:
        print("User not recognized")
```

```python
# Exam Result with Bonus Points 
marks = int(input("Marks: "))
attendance = int(input("Attendance %: "))

if marks >= 35:
    if attendance >= 75:
        if marks >= 85:
            print("Distinction")
        else:
            print("Pass")
    else:
        print("Pass but low attendance warning")
else:
    if attendance >= 90:
        print("Fail but eligible for re-test due to high attendance")
    else:
        print("Fail")
```

```python
# Marks, Attendance & Scholarship Check 
marks = int(input("Marks: "))
attendance = int(input("Attendance %: "))

if marks >= 60:
    if attendance >= 80:
        extra_curricular = input("Any extra-curricular achievements? (yes/no): ").lower()
        if extra_curricular == "yes":
            print("Scholarship awarded!")
        else:
            print("Good performance but no scholarship.")
    else:
        print("Not eligible for scholarship due to low attendance.")
else:
    print("Marks too low for scholarship consideration.")
```
